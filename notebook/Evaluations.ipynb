```python
import sys
sys.path.append("/Users/sumitkumar/LLMOps-project")

import os
import asyncio
import threading
from pathlib import Path
from multi_doc_chat.src.document_ingestion.data_ingestion import ChatIngestor
from multi_doc_chat.src.document_chat.retrieval import ConversationalRAG

DATA_PATH = "/Users/sumitkumar/LLMOps-project/data/NIPS-2017-attention-is-all-you-need-Paper.pdf"
CHUNK_SIZE, CHUNK_OVERLAP, K = 1000, 200, 5


class LocalFileAdapter:
    def __init__(self, file_path: str):
        self.path = Path(file_path)
        self.name = self.path.name

    def getbuffer(self) -> bytes:
        return self.path.read_bytes()


def _ensure_event_loop():
    try:
        asyncio.get_event_loop()
    except RuntimeError:
        asyncio.set_event_loop(asyncio.new_event_loop())


def build_rag():
    _ensure_event_loop()
    ingestor = ChatIngestor(temp_base="data", faiss_base="faiss_index", use_session_dirs=True)
    ingestor.built_retriver(
        uploaded_files=[LocalFileAdapter(DATA_PATH)],
        chunk_size=CHUNK_SIZE,
        chunk_overlap=CHUNK_OVERLAP,
        k=K,
    )
    session_id = ingestor.session_id
    rag = ConversationalRAG(session_id=session_id)
    rag.load_retriever_from_faiss(
        index_path=f"faiss_index/{session_id}",
        k=K,
        index_name=os.getenv("FAISS_INDEX_NAME", "index"),
    )
    return rag


# Built ONCE, in the main notebook thread
RAG = build_rag()
_lock = threading.Lock()


def answer_ai_report_question(inputs: dict) -> dict:
    question = inputs.get("question", "")
    if not question:
        raise ValueError("No question provided")

    _ensure_event_loop()          # safe in worker threads too
    with _lock:                   # one call at a time through the shared chain
        answer = RAG.invoke(question, chat_history=[])
    
    return {"answer": answer}
```

```python
import os
from langsmith import evaluate
from langchain_groq import ChatGroq
from openevals.llm import create_llm_as_judge
from openevals.prompts import CORRECTNESS_PROMPT

judge_llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key=os.getenv("GROQ_API_KEY"),
)

correctness_judge = create_llm_as_judge(
    prompt=CORRECTNESS_PROMPT,
    feedback_key="cot_qa",
    judge=judge_llm,
)

def qa_evaluator(inputs: dict, outputs: dict, reference_outputs: dict):
    reference = next(iter(reference_outputs.values()))
    return correctness_judge(
        inputs=inputs["question"],
        outputs=outputs["answer"],
        reference_outputs=reference,
    )

experiment_results = evaluate(
    answer_ai_report_question,
    data="GoldensData",
    evaluators=[qa_evaluator],
    experiment_prefix="test-agenticAIReport-qa-rag",
    max_concurrency=0,
    metadata={
        "variant": "RAG with FAISS and AI Engineering Report",
        "chunk_size": 1000,
        "chunk_overlap": 200,
        "k": 5,
    },
)
```

CUSTOM CORRECTNESS EVAL

```python
import os, json, re
from langsmith.evaluation import evaluate_existing
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

judge_llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key=os.getenv("GROQ_API_KEY"),
)

JUDGE_PROMPT = ChatPromptTemplate.from_messages([
    ("system",
     "You are a strict grader of question-answering systems. Compare the "
     "STUDENT ANSWER to the GOLDEN ANSWER for the QUESTION.\n\n"
     "Judge semantic and factual alignment, not wording. Score:\n"
     "1.0 = all key facts in the golden answer are present and nothing contradicts it\n"
     "0.5 = partially correct: some key facts missing, or minor inaccuracies\n"
     "0.0 = incorrect, contradicts the golden answer, or says it cannot answer\n\n"
     "Extra correct detail not in the golden answer should NOT be penalized.\n"
     "Respond with ONLY a JSON object, no other text:\n"
     '{{"reasoning": "<1-3 sentences>", "score": <0.0, 0.5, or 1.0>}}'),
    ("human",
     "QUESTION:\n{question}\n\nGOLDEN ANSWER:\n{golden}\n\nSTUDENT ANSWER:\n{answer}"),
])

judge_chain = JUDGE_PROMPT | judge_llm


def custom_correctness(inputs: dict, outputs: dict, reference_outputs: dict) -> dict:
    golden = next(iter(reference_outputs.values())) #first value of th dict in golden data
    resp = judge_chain.invoke({
        "question": inputs["question"],
        "golden": golden,
        "answer": outputs["answer"],
    })
    text = resp.content if isinstance(resp.content, str) else str(resp.content) #some model give ans in list so this line helps to cinvert the list into string
    match = re.search(r"\{.*\}", text, re.DOTALL) #This line searches the judge's reply for a JSON object and captures it
    if not match:
        raise ValueError(f"Judge returned no JSON: {text[:200]}")
    verdict = json.loads(match.group(0))
    return {
        "key": "custom_correctness",
        "score": float(verdict["score"]),
        "comment": verdict["reasoning"],
    }


experiment_results = evaluate_existing(
    "test-agenticAIReport-qa-rag-a9fe151a",
    evaluators=[custom_correctness],
)
```

```python
df = experiment_results.to_pandas()
print(df.columns.tolist())
print(df["feedback.custom_correctness"].mean())
```

ANSWER_RELEVENCE EVAL

```python
import os, json, re
from langsmith.evaluation import evaluate_existing
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

judge_llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key=os.getenv("GROQ_API_KEY"),
)

RELEVANCE_PROMPT = ChatPromptTemplate.from_messages([
    ("system",
     "You are a strict grader of question-answering systems. Decide how well "
     "the ANSWER addresses the QUESTION. Do NOT judge factual correctness, "
     "only relevance and completeness of focus.\n\n"
     "Score:\n"
     "1.0 = directly and fully addresses every part of the question, stays on topic\n"
     "0.5 = partially addresses the question: misses a part, or includes a lot of off-topic content\n"
     "0.0 = does not address the question, is off-topic, or says it cannot answer\n\n"
     "Respond with ONLY a JSON object, no other text:\n"
     '{{"reasoning": "<1-3 sentences>", "score": <0.0, 0.5, or 1.0>}}'),
    ("human", "QUESTION:\n{question}\n\nANSWER:\n{answer}"),
])

relevance_chain = RELEVANCE_PROMPT | judge_llm


def answer_relevance(inputs: dict, outputs: dict) -> dict:
    resp = relevance_chain.invoke({
        "question": inputs["question"],
        "answer": outputs["answer"],
    })
    text = resp.content if isinstance(resp.content, str) else str(resp.content)
    match = re.search(r"\{.*\}", text, re.DOTALL)
    if not match:
        raise ValueError(f"Judge returned no JSON: {text[:200]}")
    verdict = json.loads(match.group(0))
    return {
        "key": "answer_relevance",
        "score": float(verdict["score"]),
        "comment": verdict["reasoning"],
    }


experiment_results = evaluate_existing(
    "test-agenticAIReport-qa-rag-a9fe151a",
    evaluators=[answer_relevance],
)
```

```python
df = experiment_results.to_pandas()
print(df.columns.tolist())
print(df["feedback.answer_relevance"].mean())
```
